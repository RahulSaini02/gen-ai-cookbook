## Browser Use Agent using Nova Act



```python
from nova_act import NovaAct
from strands import Agent, tool
```

```python
tracking_id = 9400150105796009472614
website_url = "https://tools.usps.com/go/TrackConfirmAction_input"
```

```python
with NovaAct(starting_page=website_url) as nova:
  nova.start()
  nova.act("Search for tracking id input box")
  nova.act(f"Enter the tracking id: {tracking_id}")
  nova.act("Click on submit button")
  nova.act("wait for the results")
  nova.act("Get results of the tracking id")
  nova.act('''store the results as json as {
  "status": "<Current package status>",
  "date": "<Date of the latest update>",
  "time": "<Time of the latest update>",
  "location": "<Location of the latest update>"
  }''')
  nova.stop()
```

```python
import fire  # type: ignore

from nova_act import NovaAct


def main(record_video: bool = False) -> None:
    with NovaAct(
        starting_page="https://www.amazon.com",
        record_video=record_video,
    ) as nova:
        nova.act("search for a coffee maker")
        nova.act("select the first result")
        nova.act("scroll down or up until you see 'add to cart' and then click 'add to cart'")


if __name__ == "__main__":
    fire.Fire(main)
```

```python
%%writefile usps_track.py
from dotenv import load_dotenv
load_dotenv()

from nova_act import NovaAct

tracking_id = "9400150105796009472614"
website_url = "https://tools.usps.com/go/TrackAction"

with NovaAct(starting_page=website_url) as nova:
    nova.start()

    # Step-by-step instructions for clarity and reliability
    nova.act("Locate input field for tracking ID.")
    nova.act(f"Type {tracking_id} into the input box and click Track.")
    nova.act("Wait until text 'Status' or 'Delivered' appears on the page.")

    nova.act("Extract the most recent tracking update: status, date, time, and location.")
    nova.act(
        """Format the extracted information into the following JSON structure:
        {
            "status": "<Current package status>",
            "date": "<Date of the latest update>",
            "time": "<Time of the latest update>",
            "location": "<Location of the latest update>"
        }"""
    )

    nova.stop()
```
