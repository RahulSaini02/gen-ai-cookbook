# 🧪 Inference & Text Generation in Transformers

This notebook explains how language models generate text after training. We'll explore:
- Greedy decoding
- Beam search
- Sampling strategies (top-k, top-p, temperature)

---


## 💬 What is Inference?

Once a model is trained, **inference** refers to using it to **generate or predict output**.

For language models like GPT, this means:
> Given input like `"The future of AI is"`, what should come next?

---


## ⚙️ Greedy Decoding

- At each step, pick the **highest probability token**
- Fast but may lead to **repetitive or generic** results

```python
generator("The future of AI is", do_sample=False)
```

---


## 🎲 Sampling Strategies

Sampling allows diversity in output by adding randomness.

### 1. **Top-k Sampling**
- Pick the top **k most probable** tokens, then sample from them

### 2. **Top-p (Nucleus) Sampling**
- Pick the **smallest set of tokens** whose cumulative probability ≥ p (e.g., 0.9)

### 3. **Temperature**
- Controls randomness: higher = more random

---


````python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

print("Greedy:")
print(generator("The future of AI is", max_length=20, do_sample=False))

print("\nTop-k (k=50):")
print(generator("The future of AI is", max_length=20, do_sample=True, top_k=50))

print("\nTop-p (p=0.9):")
print(generator("The future of AI is", max_length=20, do_sample=True, top_p=0.9))

print("\nHigh Temperature (T=1.5):")
print(generator("The future of AI is", max_length=20, do_sample=True, temperature=1.5))
````

## ⚡ Speeding Up Inference: KV Caching

During generation, the model can reuse previously computed key-value pairs to avoid recomputation — especially helpful for long sequences.

This is used in frameworks like Hugging Face, vLLM, and Transformers.js for real-time chat applications.

---


## ✅ Summary

- Inference = using model to generate output from input
- Greedy is simple but deterministic
- Sampling (top-k, top-p, temperature) adds diversity
- KV caching boosts speed for long or interactive generation
