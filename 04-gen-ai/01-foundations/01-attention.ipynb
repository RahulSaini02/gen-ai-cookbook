# 🔍 Attention Mechanism - Foundation of Transformers

The most fundamental concept behind modern Large Language Models — the **Attention Mechanism**.

---


## 💡 Intuition

Imagine you're reading a sentence:
> *"The animal didn't cross the road because it was too tired."*

To understand what "it" refers to, you need to **attend** to "animal". Traditional models process sequences in order, which makes capturing long-range dependencies hard. Attention solves this by letting the model **look at all parts of a sentence at once**, deciding which words are important.

---


## 🧠 What is Self-Attention?

Given a sentence, for **each word**, the model learns:
- What am I looking for? (**Query**)
- What do I offer? (**Key**)
- What do I contain? (**Value**)

The model then compares the Query with all Keys in the sentence to decide which Values to pay attention to.

The final output is a weighted sum of all Value vectors based on how well the Query matched the Keys.

---


## 📐 Mathematical Formula

$$
\text{Attention}(Q, K, V) = \text{Softmax} \left( \frac{QK^T}{\sqrt{d_k}} \right)V
$$

Where:
- $(Q)$: Query matrix
- $(K)$: Key matrix
- $(V)$: Value matrix
- $(d_k)$: Dimension of Key vectors (used for scaling)

---


```python
import torch
import torch.nn.functional as F

# Dummy input: 3 tokens with embedding size 4
x = torch.rand(3, 4)  # (seq_len, embed_dim)

# Define weight matrices
W_q = torch.rand(4, 4)
W_k = torch.rand(4, 4)
W_v = torch.rand(4, 4)

# Linear transformations
Q = x @ W_q
K = x @ W_k
V = x @ W_v

# Scaled dot-product attention
scores = Q @ K.T / (4 ** 0.5)
weights = F.softmax(scores, dim=-1)

# Attention output
attn_output = weights @ V

print("Attention Weights:\n", weights)
print("Attention Output:\n", attn_output)
```

## 🔄 Multi-Head Attention

Instead of computing attention once, the model uses **multiple heads** in parallel. Each head learns to focus on different types of relationships — like syntax, meaning, or position.

Output from all heads is concatenated and projected again to get the final output.

---


```python
from transformers import AutoTokenizer, AutoModel
import torch

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased", output_attentions=True)

inputs = tokenizer("The cat sleeps on the mat", return_tensors="pt")
outputs = model(**inputs)

# Extract attention from the first layer
attentions = outputs.attentions
print(f"Shape of attention layer 1: {attentions[0].shape}")  # (batch_size, num_heads, seq_len, seq_len)
```

## ✅ Summary

- Attention allows models to weigh the importance of words in context.
- Self-attention lets each word look at every other word in the input.
- Multi-head attention improves model performance by offering multiple "views" of relationships.
