# Tokenization in Transformers

Before any text reaches a transformer model, it needs to be **tokenized** — converted into numerical representations that the model can understand.

This notebook covers:
- Why we tokenize
- Different tokenization techniques
- How to use tokenizers in Hugging Face

---


## 🔍 Why Tokenize?

Models cannot understand raw text. Tokenization converts words/sentences into **tokens** — often subwords — and maps them to **unique IDs**.

For example:

**Text**: `"Let's dance!"`  
**Tokens**: `["Let", "'", "s", "dance", "!"]`  
**IDs**: `[1532, 17, 18, 6540, 0]`

---


## 📦 Types of Tokenizers

### 1. **WordPiece** — used in BERT
Breaks words into subword units based on frequency.  
e.g., `"unhappiness"` → `["un", "##happi", "##ness"]`

### 2. **Byte Pair Encoding (BPE)** — used in GPT-2
Starts with characters, merges most frequent pairs.  
Good for compact vocab.

### 3. **SentencePiece** — used in T5/ALBERT
Works directly on raw text without whitespace splitting.

---

### 🆚 Comparison

| Method        | Space-Sensitive | Trainable? | Output Granularity |
|---------------|------------------|------------|---------------------|
| WordPiece     | No               | Yes        | Subwords            |
| BPE           | No               | Yes        | Subwords            |
| SentencePiece | Yes              | Yes        | Subwords/words      |

---


## 🧪 Tokenizing with Hugging Face

Let's try different tokenizers and see how they break text.


````python
from transformers import AutoTokenizer

sentence = "Let's dance!"

# BERT Tokenizer
bert_tok = AutoTokenizer.from_pretrained("bert-base-uncased")
bert_tokens = bert_tok.tokenize(sentence)
bert_ids = bert_tok.convert_tokens_to_ids(bert_tokens)

# GPT2 Tokenizer
gpt2_tok = AutoTokenizer.from_pretrained("gpt2")
gpt2_tokens = gpt2_tok.tokenize(sentence)
gpt2_ids = gpt2_tok.convert_tokens_to_ids(gpt2_tokens)

# Output
print("BERT Tokens:", bert_tokens)
print("BERT IDs   :", bert_ids)
print("GPT2 Tokens:", gpt2_tokens)
print("GPT2 IDs   :", gpt2_ids)
````

## 🏷️ Special Tokens

Transformers use special tokens to guide the model:

- `[CLS]` → Classification token (BERT)
- `[SEP]` → Separator between sentences
- `[PAD]` → Padding for equal lengths
- `<s>, </s>` → Start and end tokens (T5, BART)

Use tokenizer's built-in methods:

```python
tokenizer.cls_token, tokenizer.pad_token_id
```
