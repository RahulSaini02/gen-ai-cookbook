## 🚧 Common Limitations of Transformers

Despite their power, transformers have notable challenges:

### 1. 🧠 High Memory Usage
- Attention computation is quadratic: O(n²)
- Large batch sizes or long texts are hard to handle

### 2. 🐢 Inference Speed
- Autoregressive generation is sequential
- Slower for interactive applications

### 3. 🔍 Factual Inaccuracy
- Generative models often **hallucinate**
- Not connected to real-time knowledge

---


## ⚡ Key Optimizations

To make transformers more efficient, researchers use:

### 1. **FlashAttention**
- Computes attention with less memory and faster speed
- Used in GPT-NeoX, LLaMA 2, and others

### 2. **LoRA (Low-Rank Adaptation)**
- Fine-tunes small matrices inside a frozen model
- Saves memory and compute
- Ideal for custom use cases

```bash
pip install peft
```

### 3. **Quantization**
- Reduces weights from FP32 to INT8/INT4
- Huge boost in inference speed and size reduction

```bash
# Example with bitsandbytes
pip install bitsandbytes
```

### 4. **Distillation**
- Train a smaller model to mimic a large model
- Hugging Face supports `DistilBERT`, `TinyBERT`
